_Области Git_

__Статусы отслеживания файлов__



У каждого файла в Git может быть 1 из 4 статусов.

Untracked - новый файл

Staged - файлы, которые уже добавлены в индекс или подготовили для сохранения изменений

(Untracked может перейти в Staged)

Unmodified - файлы, которые сохранены в новой версии (когда в репозитории находится одна и та же версия файла, то состояние такого файла немодифицированный)

Modified - это unmodified файл, который модифируется, этот файл также можно перенести в staged

Эти измененные файлы в новой версии проекта становятся unmodified

__Объекты в Git__



Blob - эти объекты создаются для файла, причем для каждой версии файла создается отдельный blob и различные версии файла сохраняются в репозитории Git

Tree - эти объекты представляют собой папки, они сохраняются в репозитории Git. Папка содержат ссылка на те файлы, которые находятся в папке или подпапке

Commit - этот объект создается, когда мы сохраняем текущую версию своего проекта. Он содержит метаданные + имя и имейл автора

Annotated Tag - он создается на любом жизненном этапе нашего проекта и по ним также можно перемещаться

В папке objects хранятся хэши объектов -> каждый объект в Git имеет уникальный ID (SHA1 ХЭШ)

Первые два символа - название папки, остальные - названия файлов

__Коммит__ - указатель на дерево



При сохранении данной версии проекта Git создает 5 объектов. 2 для Tree (папок) и для 3 Blob (файлов)

Commit же будет указывать на корневую папку проекта (определенную версию)



В коммите находятся следующие данные:



Каждый следующий коммит ссылается на предыдущий

В гит есть определенный указатель __HEAD__, он ссылается на определенный коммит.



Мы видим в рабочей директории видим ту версию проекта, на которую ссылается HEAD

Мы двигаемся в истории коммитов путем передвижения указателя HEAD

Ветка - ссылка на коммит, она всегда ссылается на последний коммит
