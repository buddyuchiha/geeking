__Базовая информация про Docker__

__Docker__ - сервия для запуска приложений в контейнерах

Для прода чаще всего используют надстройку Kubernetes

Докер стоит использовать потому что:

1) Приложения запускаются в изолированной среде

Контейнер изолирован от других контейнеров и от внешней среды -> приложения не зависят от сторонних вещей

2) Легко запускать приложения на разных серверах

Берем докер-образ, переносим его с одного компьютера на другой и контейнеры при запуске на разных компьютерах будут вести себя одинаково

3) Все зависимости приложений устанавливаются внутри контейнеров

4) Легко масштабировать путем увеличения количества контейнеров

5) Очень удобно использовать в процессе разработки приложений

Контейнер - самый маленький элемент в мире Docker

обычно одно приложение = один контейнер







Клиент запускается в командной строке и с помощью клиента мы можем подключаться к службе докер (как локально, так и удаленно)

Служба отвечает за все задачи по созданию контейнера и обработке запросов от клиента

Хост это компьютер на котором запущен докер

Образ используется как основа для создания контейнеров, с одного образа можно создать много разных контейнеров.

В репозитории находятся различные версии образа

В реестре находятся различные версии репозитория (может быть как локальный, так и удаленный)

_Как работают контейнеры_



Для создания контейнеров необходим Docker Engine, который и запускает Docker Daemon

1) При создании контейнера номер 1 на жестком диске создадутся файлы, которые будут доступны только этому контейнеру.

2) У этого контейнера запустится определенный процесс (ядро здесь общее)

У контейнера будут независимые процессы и файлы

Чаще всего запускается один процесс

Если мы создаем много контейнеров с одного и того же образа -> у этих контейнеров будут общие файлы и они будут находиться только в одном месте

Докер автоматически останавливает те контейнеры, у которых нету активных процессов.

На Mac и Windows Docker работает след образом:



Запускается виртуальная машина (Docker Desktop)

__Что такое образы и их строение__



Используя один образ можно создать много контейнеров

В контейнерах есть процессы, в образах нету (есть только информация о них)

Строение образов:



Каждый слой - набор файлов

Несколько слоев используется для того, чтобы можно было переиспользовать слои в разных образах

1) Все слои read-only (когда образ создан -> в него нельзя вносить корректировки)

2) Образы можно перемещать и удалять

3) Образы хранятся в репозитории

4) Есть официальные образы и образы сообщества

Репозиторий Docker похож на Git репозиторий

2) Разные версии помечаются разными тэгами

3) Одна версия образа может иметь несколько тегов
