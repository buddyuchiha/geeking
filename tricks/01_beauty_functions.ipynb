__Приколдесные и красивые однострочные конструкции__

Распаковка

```python
x, y, z = input().strip().split()
```

Вместо цикла использовать map

```python
x, y, z = map(int, (x, y, z))
```

Использовать reduce чтобы из последовательности собрать одно значение

Вообщем сворачиваем всю последовательность, x применяется к первому и второму значению, потом к результаты первого и второго и третьего и так далее

```python
from functools import reduce 

volume = reduce(
    lambda x, y: x * y,
    map(int, input().strip().split())
)
```
