__Работа с файлами в FastAPI__

FastAPI поддерживает несколько типов работы с файлами:

* Загрузка файлов
* Скачивание файлов
* Статические файлы
* Валидация файлов

Основной компонент - UploadFile класс, где

* file - файловый объект
* filename - оригинальное имя файла
* content_type - MIME - тип
* size - размер файла

Метод: read(), write(), seek(), close()

File - функция используется для определения файловых параметров в эндпоинтах.
