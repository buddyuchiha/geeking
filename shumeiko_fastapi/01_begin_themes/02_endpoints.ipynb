__Эндпоионты__

Ручки (или эндпоинты) в FastAPI — это маршруты, которые отвечают на запросы от пользователей. Например, если пользователь отправляет запрос на GET /items, это может быть ручка, возвращающая список товаров. Каждая ручка связывается с функцией, которая определяет, что именно произойдет при запросе.
