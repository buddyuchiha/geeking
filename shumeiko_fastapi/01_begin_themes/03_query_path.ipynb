__Query и Path параметры__

_Query параметры_

Query parameters - это пары ключ-значение, которые добавляются в URL после знака вопроса ? и разделяются амперсандами &.

Синтаксис: ?key1=value1&key2=value2&key3=value3

Характеристики:

* Необязательные по своей природе
* Порядок не важен
* Используются для фильтрации, сортировки, пагинации
* Поддерживают множественные значения для одного ключа

_Path параметры_

Path parameters - переменные части URL пути, которые пользуются для идентификации конкретных ресурсов. Они являются обязательной частью маршрута.

Синтаксис: /ресурс/{параметр}/подресурс/{другой_параметр}

Характеристики:

* Обязательные по своей природе
* Часть URL пути
* Порядок важен в URL
* Используются для идентификации ресурсов
* Поддерживают типизацию и валидацию
