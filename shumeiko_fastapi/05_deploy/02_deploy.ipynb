__uvicorn & gunicorn__

Uvicorn - ASGI-сервер, предназначенный для запуска асинхронных Python веб-приложений.

Gunicorn - WSGI-сервер, используемый для запуска традиционных синхронных Python приложений.

Gunicorn управляет процессами и балансировкой нагрузки. Uvicorn workers обрабатывают асинхронные запросы.
