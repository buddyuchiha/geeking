__Асинхронность и Многопоточность__

Асинхонность основана на концепции event loop. Один поток обрабатывает множество задачи и это эффективно для I/O-bound приложений (сетевые запросы, бд)

Многопоточность использует несколько потоков выполнения. Переключение между потоками управляется операционной системой. Подходит для CPU-bound операций.

FastAPI построен на Starlette (асинхронный фреймворк) 

* Асинхронные эндпоинты - I/O операции (запросы к бд, API)
* Синхронные эндпоинты - для быстрых CPU операций
* Фоновые задачи - для длительных операций
* Многопоточность - для CPU-bound задач в фоне
