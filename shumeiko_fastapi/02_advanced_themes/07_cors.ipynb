__CORS__

CORS (Cross-Origin Resource Sharing) - это механизм безопасности браузеров, который позволяет веб-страницам делать запросы к доменам, отличным от домена самой страницы.

Данный механизм решает проблему политики одинакового источника

Для того, чтобы не было возможности с других сайтов обращаться к нашей API, Backend-разработчики должны в приложении прописать какие фронтенды могут обращаться к нашей API.

CORS всегда настраивается на бэкенде
