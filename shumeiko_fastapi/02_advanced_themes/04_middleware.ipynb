__Middleware__

Middleware - некая прослойка, которая в случае бэкенд разработки позволяет перед тем, как запрос получен, что-то с ним сделать и потом после того, как запрос обработан, но не отправлен пользователю тоже с ним что-то сделать.

Используется для:

1) Обработка запросов до достижения эндпоинтов



2) Аунтификация и авториазция:



3) Логирование и мониторинг 



4) CORS 

5) Кэширование 



6) Безопасность



По умолчанию все эндпоинты используют одни и те же middleware:



Для того, чтоб использовать middleware нужно написать декоратор @app.middleware(), которая принимает request из fastapi и call_next
