__Структура FastAPI приложения__

Единого соглашения на счет структуры проекта - нет

Все зависит от контекста 

Паттерн: __Feature-based structure (разделение по компонентам)__



* app - основная директория с кодом
* tests - модульные и интеграционные тесты
* requirements.txt - зависимости Python
* .env - переменные окружения (бд, секретные ключи)
* docker-compose.yml - конфигурация Docker

__Составные app:__

1) core - ядро приложения

_Тут хранятся фундаментальные компоненты, не зависящие от бизнес-логики_

* config.py - конфигурация (настройки, параметры, логирование)

* security.py - безопасность (аутентификация, JWT операции, шифрование/дешифрование)

2)  api - обработка HTTP запросов и маршрутизация

_Хранятся DI - функции, валидация, эндпоинты, кэширование, ротуер и т.п_

3) models - модели данных

_Хранятся ORM модели_

Чистые модели данных без бизнес-логики

4) schemas - схемы

_Хранятся pydantic схемы для валидации и сериализации_

Разделение схем для разных операций (CRUD)

5) services - сервисы

_Хранися бизнес-логика приложения_

Сервисы не знают о HTTP, работают только с моделями и схемами

6) database - база данных

_Вся логика работы с базой данных_

* base.py - базовые настройки (базовый класс, метаданные, общие миксины)

* session.py - подключение к БД 

* репозитории -  паттерн Repository для абстракции доступа к данным

7) utils - вспомогательные функции и утилиты

* работа с файлами

* кастомные валидаторы

* константы

8) tests - тесты

_Модульное и интеграционное тестирование_

__Принципы организации кода__

1) Разделение ответственности

* API - http запросы/ответы

* Services - бизнес - логика

* Repositories - доступ к данным

* Models/schemas - структура данных

2) Dependency Injection

Зависимости передаются явно

Легкое тестирование 

Гибкая конфигурация

3) Принцие единой ответственности

Каждый модуль/класс отвечает за одну конкретную задачу

4) Инверсия зависимостей 

Высокоуровневые модули не зависят от низкоуровневых

5) Легкость тестирования

Каждый компонент можно тестировать изолированно

__Примеры других архитектур__

1) Монолитная архитектура 



* Все в одном месте, прямые вызовы и простая разработка.

* Сложность тестирования, масштабирования и медленная разработка при росте

Используют для MVP, pet-проектов и прототипов

2) Модульная архитекутра

Пример - паттерн, который был разобран выше

* Хорошая масштабируемость, легкое тестирование, чиста архитектура

* Избыточность для маленьких проектов, сложнее понимать потоки данных

Чаще все используют для средних и крупных проектов.

3) Доменно-ориентированная архитектура



* Отлично отражение бизнес-логики, гибкость к изменениям требований, изоляция доменов

* Высокий порог входа, сложность реализации, больше абстракций

Используют для сложных бизнес-доменов или проектов с частыми изменениями требований

4) Гексагональная архитектура



* Полная изоляция бизнес-логики, гибкость к изменениями, независимость от фреймворков

* Высокая сложность, долгая разработка, избыточность для простых проектов

Используют для критических бизнес-систем, систем с частой сменой инфраструктуры, систем с высоким требованиям к тестированию

5) Архитектура на основе событий 



* Высокая масштабируемость, отказоустойчивость, интеграция новых компонентов

* Сложная отладка, мониторинг, высокие требования к инфраструктуре

Испльзуют для высоконагруженных и распределенных систем

6) Микросервисная архитектура



* Устойчивость к отказам, технологическая гибкость, независимоей развертывание

* Сложность отладки, нагрузки на сеть

Используют для крупных распределенных систем и для проектов в компаниях с большими командами
