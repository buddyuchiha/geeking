__Авторизация на FastAPI__

Аутентификация - процесс проверки подлинности пользователя, то есть подтверждение того, что пользователь действительно является тем, за кого себя выдает.

Пример:



Авторизация - процесс проверки прав доступа пользователя к определенным ресурсам или действиям.

Пример



JWT (JSON Web Token) - современный стандарт для создания токенов доступа, который позволяет безопасно передавать информацию между сторонами в виде компактного JSON-объекта.

Состоит из 

* Header (Заголовок) - инфа о типе токена и алгоритме шифрования
* Payload (Полезная нагрузка) - тело токена, где хранятся все данные
* Signature (Подпись) - часть, котороая обеспечивает безопасность и целостность токена

Процесс работы:

1. Аутентификация: Пользователь вводит логин/пароль в приложении.
2. Проверка и генерация: Сервер отправляет данные. Если все верно, он генерирует JWT-токен и отправляет его обратно клиенту.
3. Хранение и отправка: Клиент сохраняет токен и прикладывает его к каждому последующему запросу к серверу
4. Верификация: Сервер получает запрос, проверяет подпись JWT.
