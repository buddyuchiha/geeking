__Транзакции__ 

Транзакция - это последовательность SQL-операций, которая выполняется как единое целое. Либо все операции выполняются успешно, либо ни одна из них.

ACID - основные свойства транзакций

1) A - Atoicity (Атомарность)

Транзакция выполняется целиком или не выполняется вообще. Нет промежуточных состояний.



2) C - Consistency (Согласованность)

Транзакция переводит базу данных из одного согласованного состояния в другое. Все ограничения (constraints) соблюдаются.

3) I - Isolation (Изолированность)

Транзакции выполняются изолированно друг от друга. Параллельные транзакции не мешают друг другу.

4) D - Durability (Долговечность)

После подтверждения транзакции (COMMIT) изменения сохраняются даже при сбое системы.

__Уровни изоляции__

1. READ UNCOMITTED

Видны незакоммиченные изменения других транзакций

Возможны "грязные" чтения (dirty reads)

Самая низкая изоляция, высокая производительность

2) READ COMMITTED (по умолчанию в PostgreSQL)

Видны только закоммиченные изменения

Нет dirty reads, но возможны non-repeatable reads

Баланс между изоляцией и производительностью

3) REPEATABLE READ

Гарантирует, что повторное чтение даст те же результаты

Нет dirty reads и non-repeatable reads

Возможны phantom reads

4) SERIALIZABLE

Полная изоляция, как будто транзакции выполняются последовательно

Самая высокая изоляция, низкая производительность

Гарантирует полную консистентность

Типы блокировок:

1) Shared locks (чтение) - несколько транзакций могут читать

2) Exclusive locks (запись) - только одна транзакция может писать

Транзакции используются для финансовых операций, миграций, обработке ошибок
