__JOINS__

JOIN - операция объединения данных из двух или более таблиц на основе условия связи между ними. Это одна из самых важных и мощных возможностей SQL.

Базовые типы JOIN:

1) INNER JOIN (Внутренне соединение)

Возвращает только совпадающие записи из обеих таблиц.





2) LEFT JOIN (Левое соединение)

Все записи из левой таблицы + совпадающие из правой.





3) RIGHT JOIN (Правое соединение)

Все записи из правой таблицы + совпадающие из левой.

Все записи из правой таблицы + совпадающие из левой.



4) FULL OUTER JOIN (Полное внешнее соединение)

Все записи из обеих таблиц





5) CROSS JOIN (Перекрестное соединение)

Декартово произведение - каждая строка с каждой.





При работе с джойнами рекомендуют использовать алиасы:



INNER JOIN - когда:

Нужны только полные совпадения

Данные должны быть в обеих таблицах

95% случаев в реальных приложениях

LEFT JOIN - когда:

Нужны все записи из главной таблицы

Есть опциональные связи (например, пользователь может иметь аватарку, а может нет)

Агрегация с возможными NULL значениями

RIGHT JOIN - когда:

Редко используется на практике

Обычно переписывается как LEFT JOIN

Может быть полезен в специфических случаях

FULL JOIN - когда:

Нужно полное объединение данных

Анализ отсутствующих связей в обе стороны

Сравнение двух таблиц

CROSS JOIN - когда:

Нужно декартово произведение

Генерация всех возможных комбинаций

Очень редко в production коде
