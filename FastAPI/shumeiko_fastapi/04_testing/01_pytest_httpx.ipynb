Пирамида тестирования:



_Юнит-тесты_ - это тестирование отдельных компонентов (юнитов) приложения изолированно от других частей.

Тестируются:

* Отдельные функции
* Методы классов
* Логика преобразования данных
* Валидация

_Интеграционные тесты_ - это тестирование взаимодействия нескольких компонентов системы вместе.

Тестируются:

* Взаимодействие с базой данных
* Работа API эндпоинтов
* Интеграция с внешними сервисами
* Работца цепочки обработки

Pytest - фреймворк для тестирования Python-кода. Он позволяет писать простые и масштабируемые тесты.

Установка Pytest:

pip install pytest

Есть асинхронный

pytest-acyncio
