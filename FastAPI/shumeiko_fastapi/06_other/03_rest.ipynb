__REST__

REST (Representational State Transfer) - архитектурный стиль для построения распределенных систем, особенно веб-сервисов. 

Принципы REST:

1. Единобразие интерфейса

_Системы должна иметь единый стандарт взаимодействия_



2. Отсутствие состояния

_Каждый запрос содержит всю необходимую информацию для его обработки_



3. Кэшируемость

Ответы должны явно указывать, можно ли их кэшировать



4. Клиент-серверная архитектура

_Распределение ответственности между клиентом и сервером_





5. Слоистая система

Архитектура может состоять из нескольких слоев



6. Код по требованию

Сервер может передавать испольняемый кодклиенту
