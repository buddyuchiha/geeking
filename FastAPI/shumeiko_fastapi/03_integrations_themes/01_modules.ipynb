__Полезные библиотеки и расширения для FastAPI__




1) Шаблоны/генераторы проекта



Это библиотеки, которые зачастую нужны новичкам, чтобы строить хорошую структуру проекта

Они хорошо покажут, как разные разработчики видят структуру FastAPI



2) Валидация и сериализация данных



Pydantic сильно интегрирован в FastAPI, но еще используются датаклассы и attrs

Pydantic медленнее, чем датаклассы, но не сильно.

3) Базы данных



databases - эт надстройка или упрощение над SQLAlchemy

SQLModel интегрирует Pydantic и SQLAlchemy

4) Авторизация / Аутентификация 



AuthX настроена на Pydantic и SQLAlchemy первых версий

Лучше присматриваться к кастомной аутентификации

5) Админки



Админки - это веб интерфейс для работы с базой данных

6) Прочее



fastui - для людей, которые хотят делать небольшой фронт

fastapi-cache - для того, чтоб быстро накинуть кэш

slowapi - оч классная библиотека для ограничения кол-ва запросов к нашей api

fastapi_profiler - чтоб понять скорость работы API
