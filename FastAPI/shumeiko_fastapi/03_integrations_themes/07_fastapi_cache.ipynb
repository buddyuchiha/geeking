__Кэширование FastAPI__

Кэширование - техника хранения часто используемых данных в быстродоступном месте для ускорения доступа к ним.

Преимущества:

- Уменьшение времени отклика
- Снижение нагрузки на базу данных
- Улучшение производительности при высоких нагрузках
- Экономия вычислительных ресурсов

Необходимые библиотеки:



Redis (Remote Dictionary Server) - это база данных типа "ключ-значение" в оперативной памяти.



Установить Redis через Docker:

docker run -d -p 6379:6379 --name redis redis:latest

Методы:

* expire - установка TTL для существующего ключа

* exripeat - установки времени истечения по timestamp

* ttl - проверка оставшегося времени

* persist - удаление ttl (сделать ключ перманентным)

* setex - создание ключа с TTL в одной операции

* psetex - аналогично setex, но время в миллисекундах

Структуры данных в Redis:

1) Строки (Strings) - самый простой тип



2) Хэши (Hashes) - для объектов



expire - установка TTL для существующего ключа

3) Списки (Lists) - упорядоченные коллекции



4) Множества (Sets) - уникальные неупорядоченные элементы



5) Упорядоченные множества (Sorted Sets)



TTL (Time To Live) - время жизни ключей
