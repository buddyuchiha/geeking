__SQL Admin__

SQLAdmin - административная панель для FastAPI, которая автоматически генерирует веб-интерфейс для управления данными в вашей базе данных. 

Возможности:

* Автоматическое создание CRUD интерфейса
* Поддержка SQLAlchemy моделей
* Пагинация, фильтрация, поиск
* Аутентификация и авторизация
* Кастомизация внешнего вида и поведения

Установка:

pip intall sqladmin
