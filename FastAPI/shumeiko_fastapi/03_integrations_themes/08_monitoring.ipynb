__Мониторинг FastAPI приложения__

Мониторинг играет большую роль в обеспечении производительности, доступности и стабильности FastAPI-приложений. 

_Prometheus_ - система мониторинга с открытым исходным кодом, которая собирает метрики из вашего приложения и сохраняет их в базе данных временных рядов. Он может мониторить производительности вашего приложения и предупредит вас, когда что-то пойдет не так.

_Grafana_ - инструмент визуализации с открытым исходным кодом, который можно использовать ждля создани дашбордов для вашего приложения.

Prometheus используется для сборки метрик из приложений.

Grafana визуализирует все это.
