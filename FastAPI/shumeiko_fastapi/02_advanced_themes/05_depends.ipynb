__Depends__

Dependency Injection - это паттерн проектирования, при котором зависимости объекта передаются извне, а не создаются внутри объекта.



Принципы DI:

* Инверсия зависимостей - модули высокого уровня не должны зависеть от модулей низкого уровня
* Разделение ответственности - каждый компонент отвечает за одну задачу
* Тестируемость - легко заменять реальные зависимости mock-объектами

Depends - механизм внедрения зависимостей в FastAPI. Он нужен чтобы:

* Выносить повторяющуюся логику из маршрутов
* Управлять зависимости между компонентами
* Создавать модульные и тестируемые приложения
* Автоматически обрабатывать аутентификацию, валидацию и так далее.

Depends в FastAPI позволяет внедрять зависимости в ваши функции маршрутов (эндпоинты). Это значит, что вы можете вызывать другие функции, которые возвращают значения, необходимые для работы эндпоинта. Depends заботится о выполнении этих функций и передаче результатов в ваш эндпоинт.



_Основное правило при таком архитектурном подходе: Зависимости могут быть направлены только внутрь._

Каждый внутренний круг системы ничего не знает о внешнем.

Круг состоит из 4 слоев.

Внешний круг - бд и фреймворки.

Интерфейсы-адаптеры отвечают за преобразование данных в более удобный формат для исполнения сценариев. Грубо говоря, мост между внешний слоем и слоем бизнес-логики

Слой "сценарии" предназначен для бизнес-логики, инкапсулирует случаи использования систем. Мы также ожидаем, что изменения в этом слое не отразится на слое сущностей.

Слой-сущности это минимальная единица с чем работает приложения. Она определяется бизнес-правилами прдеприятия.
