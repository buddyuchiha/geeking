__Lifespan__

Lifespan (жизненный цикл) в FastAPI - механизм для выполнения кода при запуске и остановке приложения. 

Он позволяет:
* Инициализировать ресурсы при старте (базы данных, кэши, подключеня)
* Корректно освобождать ресурсы при завершении
* Выполнять подготовительные операции перед началом работы

Основные события жизненного цикла:

1. startup - выполняется при запуске приложения

2. shutdown - выполняется при остановке приложения
