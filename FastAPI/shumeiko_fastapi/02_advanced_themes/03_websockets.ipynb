__Websockets__

WebSocket - протокол связи, обеспечивающий двусторонний канал связи через одно TCP-соединением. В отличие от HTTP, который работает по модели запрос-ответ, WebSocket поддерживает постоянное соединение, позволяя серверу отправлять данные клиенту.
