__Валидация данных с Pydantic__

Pydantic - библиотека, которая встроена в FastAPI и используется для работы с данными. 

Она помогает проверять и преобразовывать данные, чтобы они соответствовали нужным форматам.

Модель Pydantic - класс, наследуемый от BaseModel. Поля модели определяются через аннотации типов:



* Pydantic автоматически проверяет типы данных.
* Преобразует входные данные в указанные типы (если возможно)
* Генерирует понятные ошибки валидации

В FastAPI Pydantic модели используются для:

* Валидации выходных данных (запросы)
* Сериализации выходных данных (ответы)
* Автоматической генерации документации
