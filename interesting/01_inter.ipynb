По сути интерпретатор это прослойка между кодом и оборудованием, на котором воспроизводится программа.

Чаще всего он противопоставляется компилятору

В составе интерпретатора питона есть компилятор:



Компилятор проверяет различные синтаксические проблемы (отступы скобки), но НЕ проверяет матешу и типизацию

При первом запуске проблемы/при каком-либо изменении компилятор складывает байткод в pycache

Если ничего не меняется, компилятор берет байт-кодл из pycache
