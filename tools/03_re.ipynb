__Объект re.Match, методы re.search, re.finditer, re.findall__

Существует два различных подхода обработки:



Объект Match имеет метод group(), который возвращает группу по переданному индексу, чтоб вернуть все группы -> groups()

lastindex содержит индекс последний сохраняющей группы

Чтоб узнать позиции в тексте начало и конца -> start()

Метод span() позволяет найти начальное и последнее вхождение групп по индексу

endpos() последняя позиция метода search

pos() начальный индекс поиска

re возвращает анализируемый шаблон, string анализируемую строку

Метод search():



Ключевая особенность - поиск именно __первого__ вхождения

Метод finditer():



Позволяет перебрать все вхождения

Метод findall():



Вовращает список кортежей всех вхождений
