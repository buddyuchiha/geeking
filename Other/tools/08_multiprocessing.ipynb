__Multiprocessing python__

Мультипроцессинг - это подход к параллельным вычислениям, при котором несколько процессов выполняются одновременно на разных ядрах процессора. В Python модуль multiprocessing создан для обхода ограничений GIL (Global Interpreter Lock), который мешает потокам эффективно использовать многоядерные процессоры.

Процесс - самостоятельная единица выполнения с собственным адресным пространством

GIL - глобальная блокировка интерпретатора, ограничивающая выполнение Python-кода одним потоком

Разделяемая память - механизм обмена данными между процессами

IPC (Inter-Process Communication) - методы взаимодействия процессов

Мультипроцессинг решает любые задачи, ускоряя их и распаралелливая на ядра процессора

Он создает несколько процессов (каждый процесс эт отдельная программа независимая от первой, отдельный GIL и отдельная память), взаимодействие между ними требует pickle (прерващения объектов python в байты и наоборот)

_Нет смысла создавать больше процессов, чем у нас есть ядер_

Мультипроцессинг оч похож на multithreading по написанию кода и т.п

Для работы с мультипроцессингом важно помнить, что код должен быть защищен if __name__ == '__main__':, особенно на Windows.

API принципиально похоже намногопоточность, выгодно использовать Pool, а для взаимодействия между процессами Queue и Pipe

Плюсы:
1) реальная параллельность любых задач
2) не умирает из-за одного
3) процессы не зависят друг от друга

Минусы:
1) Потребление ресурсов (памяти, процессора, времени)
2) Необходимость сериализации в pickle
3) Проблемы синхронизации (взаимодействие между процессами)
