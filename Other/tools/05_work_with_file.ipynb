_TXT_

Для открытия файла используется функция open(), она возвращает файловый объект.

Ее основные атрибуты:

1) file -> путь к файлу
2) mode -> режим открытия ('r', 'w' и т.д)
3) encoding -> кодировка

Для чтения файла используются функции:

1) read() -> прочитать весь файл



2) readline(), readlines() -> чтение построчно



3) итерация по файлу (для больших файлов лучше всего)



Для записи в файл используются функции:

1) write() -> запись строки



2) writelines() -> запись списка строк



Из некоторых советов:

Лучше использовать контекстный менджер with для работы с файлами, так как он сам закрывает их



Для работы с путями рекомендуется использовать библиотеку pathlib
