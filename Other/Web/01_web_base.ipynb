Стек TCP/IP состоит из четырех уровней, каждый из которых решает свою задачу:



1) В канальном уровне все сводится к передаче сырых битов: нулей и единиц по физическим каналам



2) В сетевом уровне данные доставляются от одной сети (домашней сети) до сервера. Данные разбиваются на пакеты



3) В транспортном уровне уже передаются данные. Также на этом уровне работают порты, они позволяют устройствам понять к какому сервису направлен трафик.



4) Ну и основной рабочий прикладной уровень
